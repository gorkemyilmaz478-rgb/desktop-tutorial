# İkinci El Araç Fiyat Tahmini
## Used Car Price Prediction

**Proje amacı:** Araçların marka, model, üretim yılı, kilometre, yakıt türü, şanzıman ve geçmiş bilgilerini kullanarak ikinci el araç fiyatını tahmin etmek.

**Karşılaştırılacak modeller:** Linear Regression ve Random Forest Regressor.

> Bu notebook, önceki çalışmadaki veri tanıtımı ve keşifsel veri analizi adımlarını koruyup veri hazırlama ve modelleme aşamalarını eklemek üzere düzenlenmiştir. Hücreleri Google Colab'da yukarıdan aşağıya çalıştırın.

## 1. Kütüphaneler ve veri setinin yüklenmesi

Google Colab dosya yükleme penceresi açıldığında `used_cars(1).csv` dosyasını seçin. Dosya adının farklı olması sorun değildir; yüklenen ilk CSV dosyası otomatik seçilir.

```python
import io
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files

uploaded = files.upload()
csv_files = [name for name in uploaded.keys() if name.lower().endswith(".csv")]

if not csv_files:
    raise FileNotFoundError("Lütfen bir CSV veri seti yükleyin.")

csv_name = csv_files[0]
veri = pd.read_csv(io.BytesIO(uploaded[csv_name]))

print("Yüklenen dosya:", csv_name)
print("Veri setinin boyutu:", veri.shape)
display(veri.head())
```

## 2. Veri setini tanıma

Veri setindeki her satır bir araç ilanını temsil eder. Hedef değişken `price` sütunudur.

```python
print("Sütunlar:")
print(veri.columns.tolist())

print("\nVeri tipleri ve dolu değer sayıları:")
veri.info()

print("\nEksik değer sayıları:")
display(veri.isna().sum().to_frame("Eksik değer"))

print("\nTekrarlanan satır sayısı:", veri.duplicated().sum())
```

## 3. Veri temizleme ve dönüştürme

`price` ve `milage` sütunları metin biçimindedir. Para birimi işaretleri, virgüller ve mil birimi temizlenerek sayısal formata dönüştürülür. Eksik kategorik değerler modelleme aşamasında eğitim verisinden öğrenilen en sık değer ile doldurulacaktır.

```python
veri_temiz = veri.copy()

veri_temiz["price"] = pd.to_numeric(
    veri_temiz["price"].astype(str).str.replace(r"[$,]", "", regex=True),
    errors="coerce"
)

veri_temiz["milage"] = pd.to_numeric(
    veri_temiz["milage"].astype(str).str.replace(r"[^0-9.]", "", regex=True),
    errors="coerce"
)

# Yaygın boş gösterimleri eksik değer olarak ele al
bos_degerler = ["", "–", "-", "N/A", "na", "None"]
for sutun in veri_temiz.select_dtypes(include="object").columns:
    veri_temiz[sutun] = veri_temiz[sutun].replace(bos_degerler, np.nan)

# Fiyatı olmayan kayıtlar hedef değişken olmadığı için çıkarılır
veri_temiz = veri_temiz.dropna(subset=["price"]).copy()

print("Temizleme sonrası boyut:", veri_temiz.shape)
print("\nSayısal sütunların özeti:")
display(veri_temiz[["model_year", "milage", "price"]].describe())
print("\nEksik değerler:")
display(veri_temiz.isna().sum().to_frame("Eksik değer"))
```

## 4. Keşifsel veri analizi ve grafikler

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(veri_temiz["price"], bins=40)
ax.set_title("Araç Fiyatlarının Dağılımı")
ax.set_xlabel("Araç fiyatı (USD)")
ax.set_ylabel("Araç sayısı")
fig.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(8, 4))
ax.scatter(veri_temiz["milage"], veri_temiz["price"], alpha=0.45)
ax.set_title("Kilometre ve Araç Fiyatı İlişkisi")
ax.set_xlabel("Kilometre (mil)")
ax.set_ylabel("Fiyat (USD)")
fig.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(8, 4))
ax.scatter(veri_temiz["model_year"], veri_temiz["price"], alpha=0.45)
ax.set_title("Model Yılı ve Araç Fiyatı İlişkisi")
ax.set_xlabel("Model yılı")
ax.set_ylabel("Fiyat (USD)")
fig.tight_layout()
plt.show()

print("Ortalama fiyat (USD):", round(veri_temiz["price"].mean(), 2))
print("Medyan fiyat (USD):", round(veri_temiz["price"].median(), 2))
print("Ortalama kilometre (mil):", round(veri_temiz["milage"].mean(), 2))
```

## 5. Makine öğrenmesi için özelliklerin belirlenmesi

Modelde marka, model, model yılı, kilometre, yakıt türü, şanzıman, kaza geçmişi ve temiz başlık bilgisi kullanılacaktır. `engine`, iç/dış renk gibi alanlar ilk model karşılaştırmasına dahil edilmemiştir; daha sonra ek özellik denemeleriyle katkıları incelenebilir.

Veri ön işleme adımları Pipeline içinde tutulur. Böylece eksik değer doldurma ve kategorik kodlama yalnızca eğitim verisinde öğrenilir; test verisine bilgi sızıntısı yapılmaz.

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

ozellikler = [
    "brand", "model", "model_year", "milage",
    "fuel_type", "transmission", "accident", "clean_title"
]
hedef = "price"

X = veri_temiz[ozellikler].copy()
y = veri_temiz[hedef].copy()

sayisal = ["model_year", "milage"]
kategorik = [s for s in ozellikler if s not in sayisal]

sayisal_islem = Pipeline(steps=[
    ("eksik_doldur", SimpleImputer(strategy="median")),
    ("olcekle", StandardScaler())
])

kategorik_islem = Pipeline(steps=[
    ("eksik_doldur", SimpleImputer(strategy="most_frequent")),
    ("one_hot", OneHotEncoder(
        handle_unknown="infrequent_if_exist",
        min_frequency=5
    ))
])

on_isleme = ColumnTransformer(transformers=[
    ("sayisal", sayisal_islem, sayisal),
    ("kategorik", kategorik_islem, kategorik)
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Eğitim kayıt sayısı:", X_train.shape[0])
print("Test kayıt sayısı:", X_test.shape[0])
```

## 6. Modellerin eğitilmesi

Fiyat değişkeninde çok yüksek değerler bulunduğu için modeller hedef değişkene `log1p` dönüşümü uygulanarak eğitilir. Tahminler değerlendirme öncesinde tekrar dolar ölçeğine çevrilir.

```python
linear_regression = TransformedTargetRegressor(
    regressor=Pipeline(steps=[
        ("on_isleme", on_isleme),
        ("model", LinearRegression())
    ]),
    func=np.log1p,
    inverse_func=np.expm1
)

random_forest = TransformedTargetRegressor(
    regressor=Pipeline(steps=[
        ("on_isleme", on_isleme),
        ("model", RandomForestRegressor(
            n_estimators=250,
            min_samples_leaf=2,
            random_state=42,
            n_jobs=-1
        ))
    ]),
    func=np.log1p,
    inverse_func=np.expm1
)

modeller = {
    "Linear Regression": linear_regression,
    "Random Forest": random_forest
}

for ad, model in modeller.items():
    print(ad, "eğitiliyor...")
    model.fit(X_train, y_train)

print("Model eğitimi tamamlandı.")
```

## 7. Model performanslarının karşılaştırılması

- **MAE:** Tahminlerin gerçek fiyattan ortalama mutlak sapmasıdır. Düşük olması tercih edilir.
- **RMSE:** Büyük hataları daha fazla cezalandırır. Düşük olması tercih edilir.
- **R²:** Modelin hedef değişkendeki değişimi açıklama düzeyini gösterir. Yüksek olması tercih edilir.

```python
sonuclar = []
tahminler = {}

for ad, model in modeller.items():
    tahmin = model.predict(X_test)
    tahmin = np.maximum(tahmin, 0)
    tahminler[ad] = tahmin

    mae = mean_absolute_error(y_test, tahmin)
    rmse = np.sqrt(mean_squared_error(y_test, tahmin))
    r2 = r2_score(y_test, tahmin)

    sonuclar.append({
        "Model": ad,
        "MAE (USD)": mae,
        "RMSE (USD)": rmse,
        "R2": r2
    })

sonuc_tablosu = pd.DataFrame(sonuclar).sort_values("MAE (USD)")
display(sonuc_tablosu.round(3))
```

## 8. Gerçek ve tahmin edilen fiyatların karşılaştırılması

```python
en_iyi_model_adi = sonuc_tablosu.iloc[0]["Model"]
en_iyi_tahmin = tahminler[en_iyi_model_adi]

fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(y_test, en_iyi_tahmin, alpha=0.5)
ust_sinir = max(float(y_test.max()), float(en_iyi_tahmin.max()))
ax.plot([0, ust_sinir], [0, ust_sinir], linestyle="--")
ax.set_title(f"Gerçek ve Tahmin Edilen Fiyatlar - {en_iyi_model_adi}")
ax.set_xlabel("Gerçek fiyat (USD)")
ax.set_ylabel("Tahmin edilen fiyat (USD)")
fig.tight_layout()
plt.show()

print("MAE değerine göre en iyi model:", en_iyi_model_adi)
```

## 9. Sonuçların yorumlanması

Bu bölüm, modeller çalıştırıldıktan sonra elde edilen gerçek metriklere göre doldurulmalıdır. Yalnızca en düşük MAE değerine bakmak yerine RMSE ve R² değerleri de birlikte değerlendirilmelidir. Veri setindeki yüksek fiyatlı uç değerler RMSE ve R² sonuçlarını etkileyebilir.
